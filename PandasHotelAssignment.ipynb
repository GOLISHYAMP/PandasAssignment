## What is the average total cost for a stay in the dataset? Not average daily cost, but total stay cost. Feel free to round this to 2 decimal points.

```python
df.columns
```

```python
df = df.rename(columns={'nights_stay': 'total_stay'})
```

```python
df.columns
```

```python
df['total_cost'] = df['total_stay'] * df['adr']
round(df['total_cost'].mean(),2)
```
